## Unique Collision Index Values

We count the number of unique `collision_index` values in each dataset.

This helps us understand the granularity and relationship between
the datasets.

A collision can have multiple vehicles and multiple casualties, so
the Vehicle and Casualty datasets are expected to contain repeated
`collision_index` values.

```python
unique_collision_counts = pd.DataFrame({
    "Dataset": ["Collision", "Vehicle", "Casualty"],
    
    "Total Rows": [
        len(collision_df),
        len(vehicle_df),
        len(casualty_df)
    ],
    
    "Unique collision_index": [
        collision_df["collision_index"].nunique(),
        vehicle_df["collision_index"].nunique(),
        casualty_df["collision_index"].nunique()
    ]
})

unique_collision_counts
```

## Collision Index Repetition

The `collision_index` is expected to repeat in the Vehicle and Casualty
datasets because one collision can involve multiple vehicles and/or
casualties.

Therefore, repeated `collision_index` values should not automatically
be treated as duplicate records.

```python
print("Collision dataset:")
print("Total rows:", len(collision_df))
print("Unique collision_index:", collision_df["collision_index"].nunique())

print("\nVehicle dataset:")
print("Total rows:", len(vehicle_df))
print("Unique collision_index:", vehicle_df["collision_index"].nunique())

print("\nCasualty dataset:")
print("Total rows:", len(casualty_df))
print("Unique collision_index:", casualty_df["collision_index"].nunique())
```

## Collision-Level Granularity Check

The Collision dataset is expected to have one record per collision.

We compare the total number of rows with the number of unique
`collision_index` values to examine whether each collision is represented
once in the raw Collision table.

```python
collision_rows = len(collision_df)
collision_unique_ids = collision_df["collision_index"].nunique()

print("Total Collision rows :", collision_rows)
print("Unique collision IDs :", collision_unique_ids)

if collision_rows == collision_unique_ids:
    print("Each collision_index appears once in the Collision dataset.")
else:
    print("Some collision_index values appear more than once.")
```

## Dataset Granularity

Based on the structure of the raw datasets:

- Collision: one row represents one collision.
- Vehicle: one row represents one vehicle involved in a collision.
- Casualty: one row represents one casualty associated with a collision
  and vehicle.

Repeated `collision_index` values in the Vehicle and Casualty datasets
are therefore expected and do not automatically indicate duplicate rows.

```python
granularity_map = pd.DataFrame({
    "Dataset": [
        "Collision",
        "Vehicle",
        "Casualty"
    ],
    
    "Granularity": [
        "One row = one collision",
        "One row = one vehicle",
        "One row = one casualty"
    ],
    
    "Main Identifier": [
        "collision_index",
        "collision_index + vehicle_reference",
        "collision_index + vehicle_reference + casualty_reference"
    ]
})

granularity_map
```
